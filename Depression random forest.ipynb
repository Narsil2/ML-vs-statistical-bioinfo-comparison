### Testing random forest

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(random_state = 22)
rf.fit(x_train,y_train)
```

```python
y_pred = rf.predict(x_test)
```

```python
accuracy = metrics.accuracy_score(y_test, y_pred)
print("Accuracy:", accuracy)
```

```python
from scipy.stats import randint
from sklearn.model_selection import RandomizedSearchCV 

x_train_study = train.drop(columns=['depr_one_hot'])
groups = train['Study']
rs_space_sgkf={'max_depth':list(np.arange(10, 100, step=10)) + [None],
              'n_estimators':np.arange(10, 500, step=50),
              'max_features':randint(1,7),
              'criterion':['gini','entropy'],
              'min_samples_leaf':randint(1,4),
              'min_samples_split':np.arange(2, 10, step=2),
          'bootstrap' : [True, False]
          }

rf_sgkf = RandomForestClassifier() 
#cv =StratifiedGroupKFold(n_splits=3)
rf_random_sgkf = RandomizedSearchCV(rf_sgkf, rs_space_sgkf, n_iter=200, scoring='accuracy', n_jobs=-1, cv=StratifiedGroupKFold(n_splits=3), verbose=3, random_state=42)
model_random_sgkf = rf_random_sgkf.fit(X = x_train_study, y = y_train, groups = groups)
```

```python
print('Best random search hyperparameters are: '+str(model_random_sgkf.best_params_))
print('Best random search score is: '+str(model_random_sgkf.best_score_))
```

```python
rf2 = RandomForestClassifier(random_state = 22, bootstrap = False, criterion = 'entropy', max_depth = 90, max_features = 3, min_samples_leaf = 1, min_samples_split = 6, n_estimators = 160)
```

#### The model has not improved with randomised search.
We try to use the best parameters from random search and the previous benchmark model

```python
from sklearn.model_selection import GridSearchCV# Create the parameter grid based on the results of random search 
param_grid = {
    'bootstrap': [True, False],
    'max_depth': [80, 90, 100, None],
    'max_features': [2,'sqrt'],
    #'criterion': ['gini','entropy'],
    'min_samples_split': [2,4,6],
    'n_estimators': [100,130,160]
}# Create a based model
rf_sgkf = RandomForestClassifier()# Instantiate the grid search model
grid_search = GridSearchCV(estimator = rf_sgkf, param_grid = param_grid, 
                          cv = StratifiedGroupKFold(n_splits=3), n_jobs = -1, verbose = 3)
model_grid_sgkf = grid_search.fit(X = x_train_study, y = y_train, groups = groups)
```

```python
print('Best grid search hyperparameters are: '+str(model_grid_sgkf.best_params_))
print('Best grid search score is: '+str(model_grid_sgkf.best_score_))
```
