```python
import dask.dataframe as dd
import pandas as pd
import numpy as np
import datatable as dt
```

```python
phenotype_df = pd.read_csv('/home/student2024/Desktop/Project_Depression/combined_dataset_phenotypes_corrected.csv')
phenotype_df.head()
#phenotype_df.shape
```

```python
methyl_dt = dt.fread('/home/student2024/Desktop/Project_Depression/combined_dataset_methyl.csv')
methyl_dt.head()
```

```python
methyl_df = methyl_dt.to_pandas()
methyl_df = methyl_df.rename(columns={'C0':'ID'})
methyl_df.head()
#methyl_df.shape
```

```python
combined_df = methyl_df.merge(phenotype_df, how='outer', on='ID')
combined_df.head()
#combined_df.shape
```

```python
from sklearnex import patch_sklearn
patch_sklearn()
```

```python
combined_df = combined_df.drop(columns=['ID','Depression'])
```

```python
from sklearn.model_selection import GroupShuffleSplit 
splitter = GroupShuffleSplit(test_size=.20, n_splits=2, random_state = 7)
split = splitter.split(combined_df, groups=combined_df['Study'])
train_inds, test_inds = next(split)

train = combined_df.iloc[train_inds]
test = combined_df.iloc[test_inds]
```

```python
x_train = train.drop(columns=['Study','depr_one_hot'])
x_test = test.drop(columns=['Study','depr_one_hot'])
y_train = train['depr_one_hot']
y_test = test['depr_one_hot']
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()

x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
from sklearn.decomposition import PCA
#pca = PCA()
#pca.fit(x_train)
```

```python
#cumsum = np.cumsum(pca.explained_variance_ratio_)
#d = np.argmax(cumsum >= 0.95) + 1
#print(d)
```

```python
pca = PCA(n_components = 857)
pca_train = pca.fit_transform(x_train)
```

```python
print(pca_train)
```

```python
from sklearn.svm import SVC
clf_svc_linear = SVC(kernel="linear",random_state=42)
clf_svc_linear.fit(pca_train,y_train)
```

```python
from sklearn.metrics import accuracy_score
pca_test = pca.transform(x_test)
print(pca_test)
```

```python
svc_linear_pred = clf_svc_linear.predict(pca_test)
svc_linear_acc = accuracy_score(y_test, svc_linear_pred)
```

```python
print(svc_linear_acc)
```

```python
clf_svc_rbf = SVC(kernel="rbf",random_state=42)
clf_svc_rbf.fit(pca_train,y_train)
svc_rbf_pred = clf_svc_rbf.predict(pca_test)
svc_rbf_acc = accuracy_score(y_test, svc_rbf_pred)
print(svc_rbf_acc)
```

```python
from sklearn.neighbors import KNeighborsClassifier
neigh = KNeighborsClassifier(n_neighbors=5)
neigh.fit(pca_train, y_train)
neigh_pred = neigh.predict(pca_test)
neigh_acc = accuracy_score(y_test, neigh_pred)
print(neigh_acc)
```
