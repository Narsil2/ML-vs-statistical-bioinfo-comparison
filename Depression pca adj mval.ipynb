```python
import dask.dataframe as dd
import pandas as pd
import numpy as np
import datatable as dt
```

```python
phenotype_df = pd.read_csv('/home/student2024/Desktop/Project_Depression/combined_dataset_phenotypes_corrected.csv')
phenotype_df.head()
#phenotype_df.shape
```

```python
methyl_dt = dt.fread('/home/student2024/Desktop/Project_Depression/combined_dataset_methyl.csv')
methyl_dt.head()
```

```python
methyl_df = methyl_dt.to_pandas()
methyl_df = methyl_df.rename(columns={'C0':'ID'})
methyl_df.head()
#methyl_df.shape
```

```python
combined_df = methyl_df.merge(phenotype_df, how='outer', on='ID')
combined_df.head()
#combined_df.shape
```

```python
from sklearnex import patch_sklearn
patch_sklearn()
```

```python
combined_df = combined_df.drop(columns=['ID','Depression'])
```

```python
from sklearn.model_selection import GroupShuffleSplit 
splitter = GroupShuffleSplit(test_size=.20, n_splits=2, random_state = 7)
split = splitter.split(combined_df, groups=combined_df['Study'])
train_inds, test_inds = next(split)

train = combined_df.iloc[train_inds]
test = combined_df.iloc[test_inds]
```

```python
x_train = train.drop(columns=['Study','depr_one_hot'])
x_test = test.drop(columns=['Study','depr_one_hot'])
y_train = train['depr_one_hot']
y_test = test['depr_one_hot']
```

```python
from sklearn.preprocessing import StandardScaler
sc = StandardScaler()

x_train = sc.fit_transform(x_train)
x_test = sc.transform(x_test)
```

```python
from sklearn.decomposition import PCA
pca = PCA(.95)
pca.fit(x_train)
pca_train = pca.transform(x_train)
pca_test = pca.transform(x_test)
```

```python
print(pca_train)
```

```python
print(pca_test)
```

```python
from sklearn.linear_model import LogisticRegression
clf = LogisticRegression(random_state=0,max_iter = 10000).fit(pca_train, y_train)
```

```python
y_pred = clf.predict(pca_test)
from sklearn.metrics import accuracy_score
acc = accuracy_score(y_test, y_pred)
print(acc)
```

```python
from sklearn.neighbors import KNeighborsClassifier
neigh = KNeighborsClassifier(n_neighbors=3)
neigh.fit(pca_train, y_train)
```

```python
neigh_pred = neigh.predict(pca_test)
acc_knn = accuracy_score(y_test, neigh_pred)
print(acc_knn)
```

```python
from sklearn.tree import DecisionTreeClassifier
dec_tree = DecisionTreeClassifier(random_state=0)
dec_tree.fit(pca_train, y_train)
```

```python
dec_pred = dec_tree.predict(pca_test)
acc_dec = accuracy_score(y_test, dec_pred)
print(acc_dec)
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(random_state = 22)
rf.fit(pca_train,y_train)
```

```python
rf_pred = rf.predict(pca_test)
acc_rf = accuracy_score(y_test, rf_pred)
print(acc_rf)
```
